FEATURE SCALING

```python
import pandas as pd
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt
```

TYPES OF FEATURE SCALING
1) STANDARDIZATION
2) NORMALIZATION - A. MIN MAX SCALER B. ROBUST SCALER ... etc





HOW TO DO STANDARDIZATION

SUPPOSE WE HAVE AGE | SALARY DATA

formula x = (xi - mean)/(standard deviation)







```python
df = pd.read_csv('Social_Network_Ads.csv')
```

```python
# df = df.iloc[:,:]
```

```python
df.sample(5)
```

TRAIN TEST SPLIT

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.drop('Purchased',axis=1),df['Purchased'],test_size=0.3,random_state=0)

X_train.shape,X_test.shape
```

STANDARD SCALER

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

#fit the scaler to the train set it will learn parameters
scaler.fit(X_train)

# transform train and test sets
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
scaler.mean_
X_train_scaled
```

```python
X_train_scaled = pd.DataFrame(X_train_scaled,columns=X_train.columns)
X_test_scaled = pd.DataFrame(X_test_scaled,columns=X_test.columns)
```

```python
np.round(X_train_scaled.describe(),1)
```

EFFECT OF SCALING

```python
fig, (ax1,ax2) = plt.subplots(ncols=2,figsize=(12,5))

ax1.scatter(X_train['Age'],X_train['EstimatedSalary'])
ax1.set_title("Before Scaling")
ax2.scatter(X_train_scaled['Age'],X_train_scaled['EstimatedSalary'],color='red')
ax2.set_title("After Scaling")
plt.show()
```

```python
fig, (ax1,ax2) = plt.subplots(ncols=2,figsize=(12,5))

#Before Scaling
ax1.set_title("Before Scaling")
sns.kdeplot(X_train['Age'],ax=ax1)
sns.kdeplot(X_train['EstimatedSalary'],ax=ax1)

#After Scaling
ax2.set_title("After Standard Scaling")
sns.kdeplot(X_train_scaled['Age'],ax=ax2)
sns.kdeplot(X_train_scaled['EstimatedSalary'],ax=ax2)
plt.show()
```

WHY SCALING IS IMPORTANT

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()
lr_scaled = LogisticRegression()
```

```python
lr.fit(X_train,y_train)
lr_scaled.fit(X_train_scaled,y_train)
```

```python
y_pred = lr.predict(X_test)
y_scaled_pred = lr_scaled.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print("Actual",accuracy_score(y_test,y_pred))
print("Scaled",accuracy_score(y_test,y_scaled_pred))
```
