```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('train.csv')
df.head()
```

A. COUNTPLOT

```python
sns.countplot(x='Pclass', data=df)
# df['Survived'].value_counts().plot(kind='bar')
```
