COLUMN TRANSFORMER

WE CAN USE THIS LIKE PREVIOUSLY WE APPLIED STUFF ON DIFFERENT COLUMNS SEPERATELY BUT USING COLUMN TRANSFORMER IT IS VERY EASY

```python
import pandas as pd 
import numpy as np 
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.read_csv('covid_toy.csv')
```

```python
df['city'].value_counts()
```

```python
from sklearn.compose import ColumnTransformer
```

```python
tranformer = ColumnTransformer(transformers=[('tnf1',SimpleImputer(),['Fever']),('tnf2',OrdinalEncoder(categories=[['Mild','Strong']]),['Cough']),('tnf3',OneHotEncoder(sparse_output=False,drop='first')['Gender','City'])],remainder='passthrough')
#remainder = 'drop' will drop the unmodified columns and 'passthrough' will keep them as it is
```
