COLUMN TRANSFORMER

WE CAN USE THIS LIKE PREVIOUSLY WE APPLIED STUFF ON DIFFERENT COLUMNS SEPERATELY BUT USING COLUMN TRANSFORMER IT IS VERY EASY

```python
import pandas as pd 
import numpy as np 
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('covid_toy.csv')
```

```python
df['city'].value_counts()
```

```python
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['has_covid']),df['has_covid'],test_size=0.2,random_state=0)
```

```python
X_train
```

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer (transformers=[
('tnf1', SimpleImputer(), ['fever']),
('tnf2', OrdinalEncoder(categories = [[ 'Mild', 'Strong' ]]), ['cough']),
('tnf3', OneHotEncoder (sparse_output=False, drop='first'),['gender', 'city'])
], remainder= 'passthrough')
#remainder = 'drop' will drop the unmodified columns and 'passthrough' will keep them as it is
```

```python
transformer.fit_transform(X_train).shape
```

```python
transformer.transform(X_test).shape
```
